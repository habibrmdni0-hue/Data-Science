```python
def kategori_umur(umur):
  if umur <= 13:
    return "anak"
  elif 13 <= umur <= 17 :
    return "remaja"
  elif umur >= 18 :
    return "dewasa"

print(kategori_umur(15))
print(kategori_umur(11))
print(kategori_umur(50))
```

```python
#Dictionary#

produk = {
    "kecap": 7000,
    "mineral": 15000,
    "susu": 10000}

print("harga produk:", produk["kecap"])

for n in produk:
    harga = produk[n]
    if harga >= 10000:
        print(f"produk diatas 10000: {n}: Rp{harga}")
```
