```python
def kategori_umur(umur):
  if umur <= 13:
    return "anak"
  elif 13 <= umur <= 17 :
    return "remaja"
  elif umur >= 18 :
    return "dewasa"

print(kategori_umur(15))
print(kategori_umur(11))
print(kategori_umur(50))
```

```python
#Dictionary#

produk = {
    "kecap": 7000,
    "mineral": 15000,
    "susu": 10000}

print("harga produk:", produk["kecap"])

for n in produk:
    harga = produk[n]
    if harga >= 10000:
        print(f"produk diatas 10000: {n}: Rp{harga}")
```

```python
produk = {
    "Kopi": 15000,
    "Roti": 8000,
    "Susu": 12000
}

for nama, harga in produk.items():
    if harga > 10000:
        print(f"{nama}: Rp{harga}")
```

```python
## Numpy & Pandas ##

#3.1#

import numpy as np

arr = np.array([1, 2, 3, 4, 5])
matriks = np.array([[1, 2], [3, 4]])

# VECTORIZATION : operasi ke SEMUA elemen sekaligus , tanpa loop ! #

print(arr * 2)
print(arr + 10)
print(arr ** 2)
print(matriks * 2)
print(matriks + 10)
print(matriks ** 2)
```

```python
print(arr.mean())
print(arr.std())
print(arr.min(), arr.max(), arr.sum())
print(np.median(arr))
```
