```python
#5.2 Distribusi & Uji Hipotesis Praktis

from scipy import stats

# Uji normalitas : apakah data berdistribusi normal ?
stat, p = stats.shapiro(df['age'].dropna())
# p > 0.05 -> data bisa dianggap normal ; p <= 0.05 -> tidak normal

print(f"Statistic: {stat:.4f}, p-value: {p}")
```

```python


# Uji-t: apakah rata-rata 2 grup BERBEDA secara signifikan?
grup_a = df[df['sex'] == 'male']['fare']
grup_b = df[df['sex'] == 'female']['fare']
stat, p = stats.ttest_ind(grup_a, grup_b)
# p <= 0.05 -> perbedaan signifikan ( bukan kebetulan )
print(f"T-test - statistic: {stat:.4f}, p-value: {p}")
```

```python
#age tidak terdistribusi normal — ini insight penting
#kalau nanti kamu masuk ke pemodelan (beberapa model butuh asumsi normalitas)
#Perbedaan fare antara laki-laki dan perempuan benar-benar signifikan, bukan kebetulan
```

```python
df.groupby('sex')['fare'].mean()
```

```python
# Chi - square : apakah 2 variabel KATEGORIK saling berhubungan ?

tabel = pd.crosstab(df['pclass'], df['sex'])
chi2, p, dof, expected = stats.chi2_contingency(tabel)
#p <= 0.05 berhubungan signifikan untuk chi, karena chi mencari 2
#Apakah 2 variabel kategorik saling berhubungan?
#chi-square membandingkan pola sebaran/frekuensi kategori.

print(f"Chi2 = {chi2:.4f}, p-value = {p:.4f}, dof = {dof}")
```

```python
pd.crosstab(df['pclass'], df['sex'], normalize='index')
#normalize='index' itu parameter di pd.crosstab() yang mengubah angka mentah (count)
#jadi proporsi/persentase per baris, supaya tiap baris jumlahnya = 1.0 (100%).
```

```python
#normalize='index'	tiap baris jumlahnya 100% (bandingkan antar kolom, dalam 1 baris)
#normalize='columns'	tiap kolom jumlahnya 100% (bandingkan antar baris, dalam 1 kolom) — misal: dari semua perempuan, berapa persen yang di kelas 1/2/3
#normalize='all' / True	seluruh tabel jumlahnya 100% (proporsi terhadap total keseluruhan)
#normalize=False (default)	angka mentah, tidak dinormalisasi
#Aturan praktis: kalau mau tahu "di dalam grup X,
#gimana sebaran Y" → pakai normalize='index' (baris = grup pembanding).
#Kalau mau tahu "di dalam grup Y,
#gimana sebaran X" → pakai normalize='columns'.
```

```python
# Korelasi Pearson : kekuatan hubungan LINEAR 2 var numerik ( -1 s.d. 1)

r, p = stats.pearsonr(df['age'], df['fare'])
# |r| > 0,7 kuat, 0,3–0,7 sedang, < 0,3 lemah
#r mendekati +1 → makin umur naik, makin fare naik juga (hubungan positif kuat)
#r mendekati -1 → makin umur naik, fare makin turun (hubungan negatif kuat)
#r mendekati 0 → hampir tidak ada hubungan linear
print(f"Korelasi Pearson - r = {r:.4f}, p-value = {p:.4f}")
```
